Fiddler
-------
The spacing of the rings is only relevant for the extra credit.

Let $g = 1/8$ be the gap, let $P_n(f)$ be the probability distribution function
of the fraction of the cylinder with a path through ring $n$.

For the first ring, $P_1(f) = \delta(f-g)$.

The probability that any path goes through the second gap is $2g$, and the probability
distribution distribution is flat for $0 < f < g$, so $P_2(f) = 2$ for $0 < f < g$ and
$P_2(f) = 0$ otherwise.

Given $f$, the fraction of paths going through the first two gaps, the probability that
any of those paths going through the third gap is $g+f$ when $f \le 1 - g$, and the
probability is 1 when $f \ge 1 - g$.

Integrating, $\int_0^{1-g} df\,P_2(f)(g+f) + \int_{1-g}^1 df\,P_2(f) 
= \int_0^g df\,2(g+f) = 3g^2 = 3/64 = 0.046875$.

Extra credit
------------
Let $a$, where $a$ is uniformly distributed between 0 and 1, be the offset between the
first gap and the second gap.

The offset at the third ring of the helix starting at 0 at the first ring and passing
through $a+g$ at the second ring is $2(a+g)$.

The offset at the third ring of the helix starting at $g$ at the first ring and passing
through $a$ at the second ring is $g+2(a-g) = 2a-g$.

So the range of offsets of helices at the third ring that pass through the first two gaps
is from $2a-g$ to $2a+2g$, or $3g$.

The probability that any helix passes through the third gap is then $g + 3g = 4g = 1/2$.
