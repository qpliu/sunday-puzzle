Shortest path
-------------
If the dimensions of the trophy is $a\times b\times c$, the
the shortest distance between opposite corners is
the minimum of $\sqrt{a^2 + x^2} + \sqrt{(b-x)^2 + c^2}$.

For a 4×207×30 trophy, the square of the minimum
appears to be 44005, and I don't see the significance of
that.

For a 4×47×30 trophy, the square of the minimum
appears to be 3365, and I don't see the significance of that.

Trying various possibilities, the best I can come up with is
6×80127×51249, giving 9047411154, which could be
read as "goat allisa".

```python
def l2(a,b,c):
    x = var("x")
    if a > b:
        a,b = b,a
    if c > b:
        c,b = b,c
    return find_local_minimum((sqrt(a^2+x^2)+sqrt((b-x)^2+c^2))^2, 0, b)
```
