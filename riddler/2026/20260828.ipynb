I'll assume random means that the probability of a total of $T$ is approximately equal to the
probability of a total of $T+n$, when $T$ is large and $n$ is small.

Let $T = 10n + k$, where $n$ is a large integer and $0 \le k \le 10$.  Let $G = \lfloor7T/10\rceil$.
We can enumerate 10 (approximately) equally probable cases.

|T    |G|
|-----|-|
|10n+0|7n|
|10n+1|7n+1|
|10n+2|7n+1|
|10n+3|7n+2|
|10n+4|7n+3|
|10n+5|7n+4|
|10n+6|7n+4|
|10n+7|7n+5|
|10n+8|7n+6|
|10n+9|7n+6|

And there are 4 unambiguous cases and 6 ambiguous cases, so the probability that the
total is unambiguous is 2/5.

Extra credit
------------
The ranks are $1/2$, $2/5$, $1/4$, $3/20$, $2/25$, $1/20$, $1/50$.  The least
common multiple of the denominators is 100, so I think there are 100 cases of
approximately equal probability to enumerate.

I note that $100n+99$ and $100(n+1)$ both give $50n+50$, $40n+40$, $25n+25$,
$15n+15$, $8n+8$, $5n+5$, $2n+2$, so those are also ambiguous, but are
counted as distinct by my code.  After accounting for that, the probability
that the set of cutoffs unambiguously determines the total is 7/10.

```python
counts = {}
for cutoffs in [tuple([round(k*rank) for rank in [1/2, 2/5, 1/4, 3/20, 2/25, 1/20, 1/50]]) for k in [0..99]]:
    counts[cutoffs] = counts[cutoffs] + 1 if cutoffs in counts else 1
(len([v for v in counts.values() if v == 1])-2)/sum(counts.values())
```
