Simulations
-----------
Running a bunch of [simulations](20260904.go) has the 5×5 board winning approximately 65% of the time.

The simulations also have the 8×8 board as having, on average, 7.25 possible bingos after randomly
removing 25 squares, while the 5×5 board has 12 possible bingos.  And a game ends with an average
of 13.6 markers on the 5×5 board and 20.9 markers on the 8×8 board.

When the 5×5 board wins, there are an average of 14.5 markers on the 5×5 board and 20.2 markers on the 8×8
board, and the 8×8 board has an average of 6.8 possible bingos.

When the 5×5 board loses, there are an average of 12.1 markers on the 5×5 board and 22.2 markers on the 8×8
boards, and the 8×8 board has an average of 8.0 possible bingos.

Fiddler
-------
There are $\binom{64}{25} = 401038568751465792$ ways to remove 25 squares from the 8×8 board.  Invariance to
90° rotations and horizontal, vertical, and diagonal reflections reduces this by a factor of no more than 8,
so it does not seem feasible to try an exact calculation.  Since 25 is odd, the symmetries reduces it by a
factor of 8.

Extra credit
------------
The 1% tolerance implies that a truncated series approximation is the way to go, such as cutting off the
calculation after a maximum of 37 or 38 markers have been placed.  However, I can't see a way around the
intractability of the 25 random squares.

Calculations
------------
One thing that can feasibly be calculated is the probability that a 5×5 board with $n$ markers and no
bingo has bingo with the $n+1$th marker, which can be an array of 21 rational numbers.  The first 4
are 0 and the last is 1.  Using brute force,

```python
b5 = [0..20]
b5[0] = 0/25
b5[1] = 0/600
b5[2] = 0/6900
b5[3] = 0/50600
b5[4] = 55/265650
b5[5] = 1100/1062380
b5[6] = 10450/3360720
b5[7] = 62700/8614980
b5[8] = 266124/18173595
b5[9] = 846944/31835504
b5[10] = 2087280/46482840
b5[11] = 4050708/56503440
b5[12] = 6224270/56823793
b5[13] = 7540124/46707252
b5[14] = 7097932/30774172
b5[15] = 5056192/15784160
b5[16] = 2611243/6034482
b5[17] = 914210/1610936
b5[18] = 194930/270949
b5[19] = 21006/24006
b5[20] = 750/750
```

Let $\beta$ iterate over all $\binom{64}{25}$ configurations of the 8×8 board.

Let $B_5(n)$ be the array above, the probability that a 5×5 board with $n$ markers and no bingo
has bingo with the $n+1$th marker.

Let $B_{8,\beta}(m)$ be the probability that an 8×8 board with $m$ markers and no bingo has bingo
with the $m+1$th marker.

Let $W_\beta(n,m)$ be the probability that the 5×5 board wins given that there are $n$ markers
and no bingo on the 5×5 board and $m$ markers and no bingo on the 8×8 board.

We have

$$ W_\beta(n,m) = \frac{25-n}{64-n-m}\left(B_5(n) + (1-B_5(n))W_\beta(n+1,m)\right)
                + \frac{39-m}{64-n-m}(1-B_{8,\beta}(m))W_\beta(n,m+1) $$

For any given $\beta$, calculating $W_\beta(0,0)$ seems feasible, since $B_{8,\beta}$ is an array
of less than 64 rational numbers.

However, the answer we want is $\sum_\beta W_\beta(0,0)/\binom{64}{25}$, and I don't see a feasible
way to calculate or approximate that.
