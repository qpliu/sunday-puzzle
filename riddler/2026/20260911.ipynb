Fiddler
-------
Let $P_n$ be the probability that Charlie
[the Cheater](https://www.democracynow.org/2026/9/11/headlines/at_gop_midterm_convention_trump_asks_supporters_to_cheat_like_hell_to_win_election) has the correct answer to the $n$th question.
For the $n+1$th (or $n-1$th) question, he has two reasonable options: randomly choose between all four choices,
or randomly choose between the three choices not matching his answer to question $n$.

If he randomly chooses between all four choices, $P_{n+1} = 1/4$.

If he randomly chooses between the three non-matching choices,
$P_{n+1} = P_n/3 + (1-P_n)\cdot2/3\cdot1/3 = 2/9 + P_n/9$, so
$P_{n+1} > 1/4$ when $P_n > 1/4$, and since $P_1 = 1$, he'll
always take this option.

The solution to the recurrence is $P_n = 1/4 + 3/4\cdot1/9^{n-1}$.

His score, on average will be $\sum_n P_n = 43/27$.

```python
maxima('load("solve_rec")')
maxima('solve_rec(P[n+1] = 2/9+P[n]/9, P[n], P[1] = 1)')
```

```python
p(n) = 1/4 + 3/4/9^(n-1)
sum([p(n) for n in [1..3]])
```

Extra credit
------------
Clearly getting answer to question 4 will yield the best average score, since the correctness probability
of guesses farther from the known answer will monotonically asymptote to 1/4.

Getting the answer to question 1 or 7 will yield $\sum_{n=1}^7 P_n = 459475/177147 \approx 2.594$.

Getting the answer to question 2 or 6 will yield $\sum_{n=1}^6 P_n + \sum_{n=2}^2 P_n = 52693/19683 \approx 2.677$.

Getting the answer to question 3 or 5 will yield $\sum_{n=1}^5 P_n + \sum_{n=2}^3 P_n = 5875/2187 \approx 2.686$.

Getting the answer to question 4 will yield $\sum_{n=1}^4 P_n + \sum_{n=2}^4 P_n = 653/243 \approx 2.687$.

```python
[sum([p(n) for n in [1,2,3,4,5,6,7]]),
 sum([p(n) for n in [1,2,2,3,4,5,6]]),
 sum([p(n) for n in [1,2,2,3,3,4,5]]),
 sum([p(n) for n in [1,2,2,3,3,4,4]])]
```

Further thoughts
----------------
If he cheats to get the first answer on a quiz with $n$ multiple choice questions when $n$
is large, then he would score $n/4 + 27/32$ on average.

If he cheats to get a middle answer, then he would score $n/4 + 15/16$ on average.
