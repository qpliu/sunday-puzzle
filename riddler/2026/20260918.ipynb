Fiddler
-------
There are 3 cyclic permutations and 3 acyclic permutations.  A swap will always either
swap from a cyclic permutation to an acyclic permutation or from an acyclic permutations
to a cyclic permutation.  And a swap can be to any of the three.

The original order is a cyclic permutation, so after any two swaps, there is a 1/3 probability
of returning to the original order, so the average number of swaps is $N = 2 + 2N/3 = 6$.

Extra credit
------------
Let $A_{a,c}$ and $C_{a,c}$ be the average number of swaps needed to visit all the permutations
from an acyclic permutation and a cyclic permutation given that $a$ acyclic permutations have
been visited and $c$ cyclic permutations have been visited.  The answer to the problem is
$C_{0,1}$, which can be found by solving a system of linear equations.  The system of equations
is $A_{a,c} = 1 + cC_{a,c}/3 + (1-c/3)C_{a,c+1}$ and $C_{a,c} = 1 + aA_{a,c}/3 + (1-a/3)A_{a+1,c}$,
where the solution gives $C_{0,1} = 171/14 \approx 12.2143$.

```python
a = [[var(f"a{ia}{ic}") for ic in [0..3]] for ia in [0..3]]
c = [[var(f"c{ia}{ic}") for ic in [0..3]] for ia in [0..3]]
eqns = []
for ia in [0..3]:
    for ic in [1..3]:
        if ia == 3 and ic == 3:
            eqns.append(a[ia][ic] == 0)
            eqns.append(c[ia][ic] == 0)
        elif ia == 3:
            eqns.append(a[ia][ic] == 1 + ic/3*c[ia][ic] + (1-ic/3)*c[ia][ic+1])
            eqns.append(c[ia][ic] == 1 + a[ia][ic])
        elif ic == 3:
            eqns.append(a[ia][ic] == 1 + c[ia][ic])
            eqns.append(c[ia][ic] == 1 + ia/3*a[ia][ic] + (1-ia/3)*a[ia+1][ic])
        else:
            eqns.append(a[ia][ic] == 1 + ic/3*c[ia][ic] + (1-ic/3)*c[ia][ic+1])
            eqns.append(c[ia][ic] == 1 + ia/3*a[ia][ic] + (1-ia/3)*a[ia+1][ic])
solve(eqns,[a[ia][ic] for ia in [0..3] for ic in [1..3]] + [c[ia][ic] for ia in [0..3] for ic in [1..3]])
```

Revisiting extra credit
-----------------------
Due to symmetry, $A_{a,c} = C_{c,a}$, so the system of equations is
$A_{a,c} = 1 + cA_{c,a}/3 + (1-c/3)A_{c+1,a}$, which has the same solution.

```python
eqns = []
for ia in [0..3]:
    for ic in [0..3]:
        if ia == 3 and ic == 3:
            eqns.append(a[ia][ic] == 0)
        elif ic == 3:
            eqns.append(a[ia][ic] == 1 + ic/3*a[ic][ia])
        else:
            eqns.append(a[ia][ic] == 1 + ic/3*a[ic][ia] + (1-ic/3)*a[ic+1][ia])
solve(eqns,[a[ia][ic] for ia in [0..3] for ic in [0..3]])
```
