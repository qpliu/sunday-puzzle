Fiddler
-------
Let the vertices of the triangle be $x_1 = r\cos\theta$, $y_1 = r\sin\theta$,
$x_2 = r\cos(\theta+2\pi/3)$, $y_2 = r\cos(\theta+2\pi/3)$,
$x_3 = r\cos(\theta+4\pi/3)$, $y_3 = r\cos(\theta+4\pi/3)$, where $r = 1/\sqrt3$.

The minimum side length is the maximum of $|x_1-x_2|$, $|x_1-x_3|$, $|x_2-x_3$,
$|y_1-y_2|$, $|y_1-y_3|$, $|y_2-y_3|$.

```python
theta = var("theta")
r = 1/sqrt(3)
x1 = r*cos(theta)
y1 = r*sin(theta)
x2 = r*cos(theta+2*pi/3)
y2 = r*sin(theta+2*pi/3)
x3 = r*cos(theta+4*pi/3)
y3 = r*sin(theta+4*pi/3)
```

```python
plot([abs(x1-x2),abs(x1-x3),abs(x2-x3),abs(y1-y2),abs(y1-y3),abs(y2-y3)], theta, 0, 2*pi/3)
```

The side length is minimized when $\theta = \pi/12$ at $(3+\sqrt3)/\sqrt{24} \approx 0.9659$.

```python
x1m = x1.substitute(theta == pi/12)
y1m = y1.substitute(theta == pi/12)
x2m = x2.substitute(theta == pi/12)
y2m = y2.substitute(theta == pi/12)
x3m = x3.substitute(theta == pi/12)
y3m = y3.substitute(theta == pi/12)
(polygon([(x2m,y2m),(x1m,y2m),(x1m,y3m),(x2m,y3m)], axes=False, color="#090")
 + polygon([(x1m,y1m),(x2m,y2m),(x3m,y3m)], color="#622")
)
```

```python
x1m-x2m, numerical_approx(x1m-x2m)
```
