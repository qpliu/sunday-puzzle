Fiddler

I'd expect twice as many samples from the faster fish, so the average is $(2\cdot2+1\cdot1)/3 = 5/3$.

Extra credit
------------
The number of samples from a fish is proportional to its speed.

Since the distribution is symmetric with respect to the sign of the velocity, we can do the
calculation with only the positive velocities.

And the calculation of the average gives
$\langle v\rangle = \int_0^\infty dv\,v^2\,e^{-v^2/2} / \int_0^\infty dv\,v\,e^{-v^2/2} = \sqrt{\pi/2} \approx 1.2533$.

```python
sigma = var("sigma")
assume(sigma > 0)
avg = integral(x^2*e^(-x^2/(2*sigma^2)), x, 0, oo)/integral(x*e^(-x^2/(2*sigma^2)), x, 0, oo)
avg,numerical_approx(avg.substitute(sigma==1))
```

Making the $\lfloor$rounds$\rceil$
----------------------------------
Since $x^2-y^2 = (x+y)(x-y)$, replacing $y$ with $iy$ gives $x^2+y^2 = (x+iy)(x-iy)$.

A similar trick can be done with
$x^4+y^4 = (x^2+iy^2)(x^2-iy^2) = (x+e^{3i\pi/4}y)(x-e^{3i\pi/4}y)(x+e^{i\pi/4}y)(x-e^{i\pi/4}y)$.

```python
x,y=var("x,y")
((x+i*y)*(x-i*y)).expand(),((x+i*sqrt(i)*y)*(x-i*sqrt(i)*y)*(x+sqrt(i)*y)*(x-sqrt(i)*y)).expand()
```

For the last case, $x^3+y^3 = (x+y)(x-(1-i\sqrt3)y/2)(x-(1+i\sqrt3)y/2)$.

```python
a,b,c = var("a,b,c")
((x+a*y)*(x+b*y)*(x+c*y)).expand()
```

```python
solve([a*b*c == 1, a*b+a*c+b*c == 0, a+b+c == 0], [a,b,c])
```

```python
((x+y)*(x-(1-i*sqrt(3))/2*y)*(x-(1+i*sqrt(3))/2*y)).expand()
```

In general, $x^n + y^n$ can be factored into $\prod_{k=1}^n(x-r_k)$ where
$r_k$ are the $n$ roots of $x^n + y^n$.

And $r_k = e^{\pi i/n+2\pi ki/n}y$.
