Fiddler

No matter where the 4th point is, there are 3 quadrilaterals, each of which
has an area that is area of the equilateral triangle minus the area of one
of the triangles formed by the 4th point and 2 of the vertices of the equilateral
triangle.  And there are 3 of those triangles, for which the average area is
1/3 the area of the equilateral triangle, so the average area of the quadrilaterals
is 2/3 the area of the equilateral triangle, which is $2/3\cdot\sqrt3/4 = 1/\sqrt{12}$.

Extra credit
------------
Since the areas of the three triangles sum to the area of the equilateral triangle, the area of
middle sized triangle must be 1/3 the area of the equilateral triangle, or
$1/3\cdot\sqrt3/4 = 1/\sqrt{48}$.

Let the vertices of the equilateral triangle be at $(-1/2,0)$, $(1/2,0)$, and $(\sqrt3/2,0)$,
and let the 4th point be at $(x,y)$.

The area of the bottom triangle is $y/2$.  When the area of the bottom triangle is $1/\sqrt{48}$,
that is the line $y = 1/\sqrt{12}$.  The segment of the line within the triangle is from $x = -1/3$
to $x = 1/3$, so its length is $2/3$.

Similarly for when the areas of the left triangle and the right triangle are $1/\sqrt{48}$.

The total length of the locus is $2$.

Visually, the locus consists of the segments within the triangle of the three lines passing through
the center of the triangle parallel to each of the sides of the triangle.

```python
solve(sqrt(3)/2 - sqrt(3)*x == 1/sqrt(12), x)
```
