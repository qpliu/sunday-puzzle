Let $p_0, p_1, p_2, p_3, p_4$ be the probabilities that a round scores $0, 1, \sqrt2, \sqrt3, \sqrt6$,
where $p_0+p_1+p_2+p_3+p_4 = 1$.  After $n$ rounds, the average score should be
$n(p_1 + p_2\sqrt2 + p_3\sqrt3 + p_4\sqrt6)$, so, on average, I expect to play
$1/(p_1 + p_2\sqrt2 + p_3\sqrt3 + p_4\sqrt6)$ rounds per point.  [Simulations](20261030.go) say that it
is approximately 137 rounds per point.

Let $p_1 = p_{1T} + p_{1B} + p_{1L} + p_{1D}$, where $p_{1T}$ is the probability that the connector
is the yellow connector connecting the two top points and scores, $p_{1B}$ is the probability that the
connector is the yellow connector connecting the bottom two points and scores, $p_{1L}$ is the probability
that the connector is the yellow connector connecting the left two points and scores, and $p_{1D}$ is the
probability that the connector is the yellow connector connecting the upper-left point to the lower-right
point.  I'm arbitrarily choosing that diagrams with connectors connecting the two right points or with
connectors connecting the upper-right and lower left points are not in the list, being equivalent
reflections of diagrams in the list.

Calculating $p_{1T}$, the probability that this is the connector is $1/24$.  The probability that the
connected points match a diagram in the list is $28/81$, since the two letters must be different and half
of them have the letters at the right points.  The probability that the two secret points match
is $1/4\cdot1/4 = 1/16$.

Similarly for $p_{1B}$, $p_{1L}$, and $p_{1D}$, noting that $p_{1L} = p_{1D}$.

Doing the same exercise for $p_2$, $p_3$, and $p_4$, then combining the results gives the
answer of $\frac{31104}{3\sqrt6 + 42\sqrt3 + 48\sqrt2 + 79} \approx 137.035999189566$.
