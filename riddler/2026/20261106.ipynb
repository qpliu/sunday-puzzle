Fiddler
-------
The probability that 6 is not rolled is $5^N/6^N$.  The probability that 1 is not rolled
is $5^N/6^N$.  The probability that neither 1 nor 6 is rolled is $4^N/6^N$, which would
be double-counted if the first two probabilities were added.  Thus, the probability that
the range is less than 5 is $(2\cdot5^N - 4^N)/6^N$.

When $N = 6$, the probability that the range is less than 5 is $13577/23328 \approx 0.5820$.

When $N = 7$, the probability that the range is less than 5 is $23311/46656 \approx 0.4996$.

So the answer is 7.

```python
p(n) = (2*5^n-4^n)/6^n
p(6), numerical_approx(p(6)), p(7), numerical_approx(p(7))
```

Let $\epsilon = 1/6^{1000}$.

The probability that 6000 is not rolled is $(1-\epsilon)^N$.  The probability that 1000 is not rolled
is $(1-\epsilon)^N$.  The probability that neither is rolled is $(1-2\epsilon)^N$.  The probability
that the range is less than 5000 is $2(1 - \epsilon)^N - (1 - 2\epsilon)^N$.

Let $x = \epsilon N$.  Then, the minimum $x$ satisfies $2(1 - x/N)^N - (1 - 2x/N)^N = 1/2$.

Since $e^x = \lim_{N\rightarrow\infty}\left(1 + \frac xN\right)^N$, the first approximation
of $x$ satisfies $2e^{-x} - e^{-2x} = 1/2$, which is a quadratic equation for $e^{-x}$ for which the
roots are $e^{-x} = 1 \pm 1/\sqrt2$.  Since $x > 0$, choose the root that is less than 1.

So the minimum $N \approx - \log(1-1/\sqrt2)/\epsilon \approx 1.7395\times10^{778}$.

```python
-numerical_approx(log(1-1/sqrt(2))*6^1000)
```
